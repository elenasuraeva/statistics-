```python
fruits = [
    "Mango", "Mango", "Cherry", "Banana", "Orange", "Banana", "Banana", "Cherry", "Apple", "Kiwi",
    "Orange", "Apple", "Mango", "Mango", "Orange", "Peach", "Kiwi", "Cherry", "Mango", "Peach",
    "Orange", "Orange", "Orange", "Peach", "Banana", "Banana", "Orange", "Cherry", "Cherry", "Cherry",
    "Peach", "Kiwi", "Peach", "Kiwi", "Banana", "Peach", "Banana", "Orange", "Orange", "Apple",
    "Cherry", "Orange", "Mango", "Orange", "Orange", "Apple", "Orange", "Orange", "Apple", "Orange",
    "Peach", "Banana", "Mango", "Peach", "Orange", "Apple", "Mango", "Banana", "Orange", "Apple",
    "Mango", "Kiwi", "Banana", "Apple", "Banana", "Orange", "Banana", "Orange", "Apple", "Orange",
    "Kiwi", "Cherry", "Orange", "Apple", "Banana", "Orange", "Apple", "Banana", "Peach", "Orange",
    "Peach", "Cherry", "Mango", "Apple", "Apple", "Banana", "Banana", "Banana", "Cherry", "Apple",
    "Mango", "Apple", "Orange", "Banana", "Orange", "Apple", "Apple", "Kiwi", "Mango", "Kiwi"
]
```

```python
from collections import Counter

counts = Counter(fruits)

print(counts)
```

```python
import matplotlib.pyplot as plt

plt.bar(counts.keys(), counts.values())
plt.xlabel("Fruit")
plt.ylabel("Frequency")
plt.title("Frequency of Fruits")
plt.show()
```
